# 🧮 Matrix Multiplication in Python

---

## 📌 Section 1: Square Matrix Multiplication (2 × 2)
Matrix multiplication between two square matrices $A$ and $B$:
$$C[i][j] = \sum_{k=0}^{n-1} A[i][k] \times B[k][j]$$

### 🔹 Method 1A: Traditional 3-Loop Iterative Approach
- **Time Complexity:** $\mathcal{O}(n^3)$
- **Mechanism:** Three nested loops iterating over rows of $A$ ($i$), columns of $B$ ($j$), and the common dimension ($k$).


```python
a = [
    [1, 2],
    [3, 4]
]

b = [
    [5, 6],
    [7, 8]
]

c = [
    [0, 0],
    [0, 0]
]

for i in range(2):
    for j in range(2):
        for k in range(2):
            c[i][j] += a[i][k] * b[k][j]

print("Matrix A:", a)
print("Matrix B:", b)
print("Multiplication of A and B:")
for row in c:
    print(row)
```

---
### 🔹 Method 1B: Pythonic Approach using `zip(*b)` and List Comprehension
- **Key Insight:** `*b` unpacks rows of $b$, and `zip(*b)` groups elements by column, effectively **transposing** matrix $b$.
- `sum(x * y for x, y in zip(row_a, col_b))` computes the dot product of a row and column in a single expressive expression.


```python
a = [
    [1, 2],
    [3, 4]
]

b = [
    [5, 6],
    [7, 8]
]

# zip(*b) transposes matrix b into columns
c = [[sum(x * y for x, y in zip(row_a, col_b)) for col_b in zip(*b)] for row_a in a]

print("Multiplication of A and B:")
for row in c:
    print(row)
```

## 📌 Section 2: Asymmetrical (Rectangular) Matrix Multiplication

> **Mathematical Requirement:**
> To multiply matrix $A$ of shape $(m \times k)$ with matrix $B$ of shape $(k \times n)$:
> - The number of **columns in $A$** must equal the number of **rows in $B$** ($k$).
> - The resulting matrix $C$ will have shape **$(m \times n)$**.

Here:
- Matrix $A$ is $2 \times 3$ ($m=2, k=3$)
- Matrix $B$ is $3 \times 2$ ($k=3, n=2$)
- Result Matrix $C$ is $2 \times 2$ ($m=2, n=2$)

### 🔹 Method 2A: Dynamic Nested Loops
Dynamically computes dimensions using `len(A)`, `len(B[0])`, and `len(B)` without hardcoding sizes.


```python
A = [[1, 2, 3], [4, 5, 6]]          # 2x3
B = [[7, 8], [9, 1], [2, 3]]          # 3x2
C = [[0] * len(B[0]) for _ in range(len(A))]
for i in range(len(A)):
    for j in range(len(B[0])):
        for k in range(len(B)):
            C[i][j] += A[i][k] * B[k][j]
print(C)
```

---
### 🔹 Method 2B: One-Line List Comprehension for Asymmetric Matrices
Generalizes the `zip(*B)` transposition technique to arbitrary rectangular dimensions.


```python
A = [[1, 2, 3], [4, 5, 6]]
B = [[7, 8], [9, 1], [2, 3]]
# zip(*B) effectively turns the columns of B  into rows so we can easily loop over them.
C = [[sum(x * y for x, y in zip(row_a, col_b)) for col_b in zip(*B)] for row_a in A]
print(C)
```

---
### 🔹 Method 2C: High-Performance Multiplication with NumPy
- Uses NumPy's highly optimized, vectorized BLAS backend.
- The `@` operator (introduced in Python 3.5 via PEP 465) provides clean, readable linear algebra syntax.


```python
# with numpy
import numpy as np

# A is 2x3, B is 3x2
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

B = np.array([
    [7, 8],
    [9, 1],
    [2, 3]
])

# Multiply using the @ operator
C = A @ B

print("Result Matrix C:\n", C)
```

## 📊 Comparison Summary

| Method | Approach | Readability | Performance | Best Used For |
| :--- | :--- | :--- | :--- | :--- |
| **Nested Loops** | $\mathcal{O}(m \cdot k \cdot n)$ | Explicit & Clear | Slower in pure Python | Educational & algorithm demonstration |
| **`zip(*B)` Comprehension** | Pure Pythonic | Concise & Functional | Moderate | Quick scripts without dependencies |
| **NumPy (`@`)** | Vectorized (C/Fortran) | Minimalist & Industry Standard | Extremely Fast | Data Science, ML & Production |
